```python
# NUMPY & SCIPY FOR DESCRIPTIVE STATS IN EDA

import numpy as np
import scipy
from scipy import stats
```

```python
# DESCRIPTIVE STATS IN EDA

def eda_selected(df, columns, trim_prop=0.1):
    results = []
    for col in columns:
        if col not in df.columns:
            print(f"{col} not found.")
            continue
        data = df[col].dropna().values

        # Parametric Central Tendency
        mean = np.mean(data)

        # Non-Parametric Central Tendency
        median = np.median(data)
        mode_val = stats.mode(data, keepdims=True).mode[0]  # Mode

        # Trimmed Parametric Central Tendency
        trimmed_mean = stats.trim_mean(data, proportiontocut=trim_prop)

        # Parametric Central Dispersion
        mean_abs_dev = np.mean(np.abs(data - mean))
        variance = np.var(data, ddof=1)
        std_dev = np.std(data, ddof=1)

        # Non-Parametric Central Dispersion
        mean_median_abs_dev = np.mean(np.abs(data - median))

        # Trimmed Parametric Central Dispersion
        sorted_data = np.sort(data)
        trim_count = int(len(data) * trim_prop)
        trimmed_data = sorted_data[trim_count: len(data)-trim_count]
        trimmed_var = np.var(trimmed_data, ddof=1)
        trimmed_std = np.sqrt(trimmed_var)

        # Non-Central Dispersion (Non-Parametric)
        range_val = np.ptp(data)
        percentile_25 = np.percentile(data, 25)
        percentile_75 = np.percentile(data, 75)
        iqr = percentile_75 - percentile_25

        # Parametric Distribution
        param_skew = stats.skew(data, bias=False)
        param_kurt = stats.kurtosis(data, fisher=False, bias=False)

        # Non-Parametric Distribution
        Q1 = np.percentile(data, 25)
        Q2 = np.median(data)
        Q3 = np.percentile(data, 75)
        bowley_skew = (Q3 + Q1 - 2*Q2)/(Q3 - Q1) if (Q3 - Q1)!=0 else np.nan
        P12_5 = np.percentile(data, 12.5)
        P37_5 = np.percentile(data, 37.5)
        P62_5 = np.percentile(data, 62.5)
        P87_5 = np.percentile(data, 87.5)
        moors_kurt = ((P87_5 - P62_5) + (P37_5 - P12_5)) / (Q3 - Q1) if (Q3 - Q1)!=0 else np.nan

        # Trimmed Parametric Distribution
        trimmed_skew = stats.skew(trimmed_data, bias=False)
        trimmed_kurt = stats.kurtosis(trimmed_data, fisher=False, bias=False)

        results.append({
            "Column": col,
            "Mean": mean,
            "Median": median,
            "Mode": mode_val,
            "Trimmed Mean": trimmed_mean,
            "Mean Absolute Deviation": mean_abs_dev,
            "Variance": variance,
            "Std Dev": std_dev,
            "Mean Absolute Deviation (from Median)": mean_median_abs_dev,
            "Trimmed Variance": trimmed_var,
            "Trimmed Std Dev": trimmed_std,
            "Range": range_val,
            "25th Percentile": percentile_25,
            "75th Percentile": percentile_75,
            "IQR": iqr,
            "Skewness": param_skew,
            "Kurtosis": param_kurt,
            "Bowley Skewness": bowley_skew,
            "Moors Kurtosis": moors_kurt,
            "Trimmed Skewness": trimmed_skew,
            "Trimmed Kurtosis": trimmed_kurt
        })
    return pd.DataFrame(results)

columns_to_analyze = ["Baseline MAND"]
eda_summary = eda_selected(df, columns_to_analyze)
print(eda_summary)
```

```python
# MATPLOTLIB & SEABORN FOR PLOTS IN EDA

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# DISTRIBUTION PLOT IN EDA

def plot_eda(df, columns, bin_width=5, kde_bw=0.5):
    for col in columns:
        if col not in df.columns:
            print(f"{col} not found.")
            continue
        data = df[col].dropna().values
        min_val = np.floor(data.min())
        max_val = np.ceil(data.max())
        bins = np.arange(min_val, max_val + bin_width, bin_width)

        # Frequency Table
        freq, bin_edges = np.histogram(data, bins=bins)
        freq_table = pd.DataFrame({
            "Bin Start": bin_edges[:-1],
            "Bin End": bin_edges[1:],
            "Frequency": freq
        })
        print(f"\nFrequency Table for {col}:")
        print(freq_table)

        # Histogram
        plt.figure(figsize=(8, 5))
        plt.hist(data, bins=bins, edgecolor='black')
        plt.title(f"Histogram of {col}")
        plt.xlabel("Value")
        plt.ylabel("Frequency")
        plt.show()

columns_to_plot = ["Baseline TACT"]
plot_eda(df, columns_to_plot, bin_width=5, kde_bw=0.5)
```

```python
# DISTRIBUTION PLOTS IN EDA (VIA SIMULATIONS)

# Simulations
distributions = {
    "Single trial, single outcome (Certainty)": lambda: np.array([1]*1000),
    "Single trial, two outcomes (Bernoulli)": lambda: stats.bernoulli(p=0.3).rvs(1000),
    "Single trial, multiple outcomes (Categorical)": lambda: stats.rv_discrete(values=([1,2,3], [0.2,0.5,0.3])).rvs(1000),
    "Single trial, unlimited outcomes (Poisson)": lambda: stats.poisson(mu=3).rvs(1000),
    "Single trial, infinite outcomes (Normal)": lambda: stats.norm(loc=0, scale=1).rvs(1000),
    "Multiple trials, single outcome (Uniform)": lambda: stats.uniform(loc=0, scale=1).rvs(1000),
    "Multiple trials, two outcomes (Binomial)": lambda: stats.binom(n=10, p=0.4).rvs(1000),
    "Multiple trials, multiple outcomes (Multinomial)": lambda: np.random.multinomial(n=10, pvals=[0.2,0.5,0.3], size=1000),
    "Multiple trials, unlimited outcomes (Poisson)": lambda: stats.poisson(mu=5).rvs(1000),
    "Multiple trials, infinite outcomes (Normal)": lambda: stats.norm(loc=0, scale=2).rvs(1000)
}

# Distribution Plots
def plot_distribution(name, dist_obj):
    data = dist_obj()
    if name == "Multiple trials, multiple outcomes (Multinomial)":
        plt.figure(figsize=(6,4))
        for i in range(data.shape[1]):
            plt.hist(data[:, i], bins=np.arange(data[:, i].min(), data[:, i].max()+2)-0.5, alpha=0.5, label=f'Category {i+1}')
        plt.legend()
        plt.title(name)
        plt.xlabel('Value')
        plt.ylabel('Frequency')
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.show()
        return

    if "Normal" in name:
      fig, ax = plt.subplots(1, 3, figsize=(18,4))

      # Histograms
      ax[0].hist(data, bins=30, alpha=0.7, color='skyblue', edgecolor='black')
      ax[0].set_title(f'{name} Histogram')
      ax[0].set_xlabel('Value')
      ax[0].set_ylabel('Frequency')
      ax[0].grid(True, linestyle='--', alpha=0.6)

      # Density plots
      sns.kdeplot(data, ax=ax[1], fill=True, color='orange')
      ax[1].set_title(f'{name} Density Plot')
      ax[1].set_xlabel('Value')
      ax[1].set_ylabel('Density')
      ax[1].grid(True, linestyle='--', alpha=0.6)

      # Q-Q Plots
      stats.probplot(data, dist="norm", plot=ax[2])
      ax[2].set_title(f'{name} Q-Q Plot')
      plt.tight_layout()
      plt.show()
      return

    plt.figure(figsize=(6,4))
    plt.hist(data, bins=30, alpha=0.7, color='skyblue', edgecolor='black')
    plt.title(name)
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()

for name, dist in distributions.items():
    plot_distribution(name, dist)
```

```python
# NUMERICAL VS NUMERICAL EDA

# Scatterplot
columns = ["Baseline MAND", "Baseline TACT", "Baseline INTRA"]
sns.pairplot(df[columns], kind='scatter', diag_kind='hist', plot_kws={'alpha':0.6, 's':50, 'edgecolor':'k'})
plt.suptitle("Scatter Plots for Baseline Columns", y=1.02)
plt.show()

# Hexagonal Binning, Heatmap & Contour Plot
def baseline_hexbin_contour_heatmap(df):
    cols = ["Baseline MAND", "Baseline TACT", "Baseline INTRA"]

    # Hexbin
    plt.figure(figsize=(15,4))
    for i, (x, y) in enumerate([("Baseline MAND", "Baseline TACT"),
                                ("Baseline MAND", "Baseline INTRA"),
                                ("Baseline TACT", "Baseline INTRA")], 1):
        plt.subplot(1, 3, i)
        plt.hexbin(df[x], df[y], gridsize=30, cmap="Blues", mincnt=1)
        plt.colorbar(label="Counts")
        plt.xlabel(x)
        plt.ylabel(y)
        plt.title(f"Hexbin: {x} vs {y}")
    plt.tight_layout()
    plt.show()

    # Heatmap
    plt.figure(figsize=(15,4))
    for i, (x, y) in enumerate([("Baseline MAND", "Baseline TACT"),
                                ("Baseline MAND", "Baseline INTRA"),
                                ("Baseline TACT", "Baseline INTRA")], 1):
        plt.subplot(1, 3, i)
        hb = plt.hist2d(df[x], df[y], bins=30, cmap="viridis")
        plt.colorbar(hb[3], label="Counts")
        plt.xlabel(x)
        plt.ylabel(y)
        plt.title(f"Heatmap: {x} vs {y}")
    plt.tight_layout()
    plt.show()

    # Contour Plot
    plt.figure(figsize=(15,4))
    for i, (x, y) in enumerate([("Baseline MAND", "Baseline TACT"),
                                ("Baseline MAND", "Baseline INTRA"),
                                ("Baseline TACT", "Baseline INTRA")], 1):
        plt.subplot(1, 3, i)
        sns.kdeplot(
            x=df[x], y=df[y],
            fill=False, cmap="Reds", levels=10, thresh=0
        )
        plt.xlabel(x)
        plt.ylabel(y)
        plt.title(f"Contour: {x} vs {y}")
    plt.tight_layout()
    plt.show()

baseline_hexbin_contour_heatmap(df)
```

```python
## CATEGORICAL VS NUMERICAL EDA

# Boxplot
def boxplot_columns(df, columns):
    valid_cols = [col for col in columns if col in df.columns]
    if not valid_cols:
        print("No valid columns found.")
        return
    data_to_plot = [df[col].dropna() for col in valid_cols]
    plt.figure(figsize=(10, 6))
    plt.boxplot(data_to_plot, labels=valid_cols, patch_artist=True,
                medianprops=dict(color='black'),
                boxprops=dict(facecolor='lightblue', color='blue'),
                whiskerprops=dict(color='blue', linestyle='--'),
                capprops=dict(color='blue'),
                flierprops=dict(marker='o', markerfacecolor='red', markersize=5, linestyle='none'))
    plt.title("Boxplot of Selected Columns")
    plt.ylabel("Value")
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
columns_to_boxplot = ["Baseline MAND", "Baseline TACT", "Baseline INTRA"]
boxplot_columns(df, columns_to_boxplot)

# Violin Plot
def baseline_violin_plot(df):
    cols = ["Baseline MAND", "Baseline TACT", "Baseline INTRA"]
    df_melted = df[cols].melt(var_name="Baseline", value_name="Value")
    plt.figure(figsize=(8,6))
    sns.violinplot(x="Baseline", y="Value", data=df_melted, palette="Pastel1")
    plt.title("Violin Plot for Baseline MAND, TACT, INTRA")
    plt.ylabel("Values")
    plt.show()
baseline_violin_plot(df)
```

```python
# MULTIVARIABLE EDA

def baseline_facet_hexbin(df):
    cols = ["Baseline MAND", "Baseline TACT", "Baseline INTRA"]
    support_order = ['Level 1', 'Level 2', 'Level 3']
    support_cat = pd.Categorical(df['Support Level'], categories=support_order, ordered=True)
    n_cols = len(cols)
    plt.figure(figsize=(5*n_cols, 4))
    for i, col in enumerate(cols, 1):
        plt.subplot(1, n_cols, i)
        plt.hexbin(support_cat.codes, df[col], gridsize=20, cmap='Blues', mincnt=1)
        plt.colorbar(label='Counts')
        plt.xticks(ticks=range(len(support_order)), labels=support_order, rotation=45)
        plt.xlabel('Support Level')
        plt.ylabel(col)
        plt.title(f'Hexbin: {col} vs Support Level')
    plt.tight_layout()
    plt.show()
baseline_facet_hexbin(df)
```
